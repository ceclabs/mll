```python
import numpy as np
import pandas as pd
from sklearn import svm, datasets
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
# Load CSV
train_df = pd.read_csv(r"fashion_mnist_train_5000.csv")
test_df = pd.read_csv(r"fashion_mnist_test_1000.csv")
```

```python
# Separate features and labels
X_train = train_df.drop('label', axis=1).values
y_train = train_df['label'].values

X_test = test_df.drop('label', axis=1).values
y_test = test_df['label'].values

print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# -------- Linear SVM --------
linear_model = svm.SVC(kernel='linear').fit(X_train, y_train)
y_pred_linear = linear_model.predict(X_test)

# -------- Polynomial SVM --------
poly_model = svm.SVC(kernel='poly', degree=3).fit(X_train, y_train)
y_pred_poly = poly_model.predict(X_test)

# -------- RBF SVM --------
rbf_model = svm.SVC(kernel='rbf').fit(X_train, y_train)
y_pred_rbf = rbf_model.predict(X_test)

# Accuracy
print("Linear SVM Accuracy:", accuracy_score(y_test, y_pred_linear))
print("Polynomial SVM Accuracy:", accuracy_score(y_test, y_pred_poly))
print("RBF SVM Accuracy:", accuracy_score(y_test, y_pred_rbf))
```
